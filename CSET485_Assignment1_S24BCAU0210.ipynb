```python
import warnings; warnings.filterwarnings('ignore')
import os, numpy as np, pandas as pd
import matplotlib.pyplot as plt
import statsmodels.api as sm
from scipy import stats
from sklearn.model_selection import train_test_split, StratifiedKFold, cross_validate
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LinearRegression, LogisticRegression
from sklearn.metrics import confusion_matrix, accuracy_score, roc_auc_score

SEED = 210                      # last 3 digits of S24BCAU0210
pd.set_option('display.width', 200); pd.set_option('display.max_columns', 50)
np.random.seed(SEED)
URL = 'https://github.com/propublica/compas-analysis/raw/master/compas-scores-two-years.csv'
LOCAL = 'compas-scores-two-years.csv'
```

## 0. Load data

```python
raw = pd.read_csv(LOCAL) if os.path.exists(LOCAL) else pd.read_csv(URL)
print('Raw shape:', raw.shape)
print('Columns with missing values:'); print(raw.isna().sum()[raw.isna().sum()>0])
```

## 1. Data Preprocessing (documentation)

**Missing values.** Many columns are partly or entirely empty (e.g. `violent_recid` is 100% missing, `c_arrest_date` ~84%, `r_*` columns are missing for people who did not re-offend, which is structural, not an error). None of those columns are used as predictors. Rows are filtered with the standard ProPublica rules: (i) `days_b_screening_arrest` within Â±30 days (screening and arrest must be linked to the same case, which also removes the 307 rows with missing values there), (ii) `is_recid != -1` (no valid recidivism info), (iii) `c_charge_degree != 'O'` (ordinary traffic offences), (iv) `score_text != 'N/A'`. All model features have 0 missing values after this step, so no imputation was needed.

**Categorical encoding.** `sex` â†’ `male` (1/0); `c_charge_degree` â†’ `felony` (F=1, M=0); `race` â†’ one-hot with Caucasian as the reference (`race_African_American`, `race_Hispanic`, `race_Other`; Asian and Native American are merged into Other because they have only 32 and 18 rows).

**Outliers.** `priors_count` is heavily right-skewed; it is winsorized at the 99th percentile of the *training set*. The three juvenile count columns are summed into `juv_total`. Age is plausible (18â€“96) and left untouched.

**Feature scaling.** Continuous features (`age`, `priors_count`, `juv_total`) are standardized with `StandardScaler` fitted on the *training set only* for Parts 1, 2.3, 2.4 (regularisation-free logistic regression is scale-invariant in its predictions, so this does not change predictions). For the coefficient-interpretation analyses (2.1, 2.2) raw units are used so a coefficient reads "per year of age" / "per prior".

**Targets.** Model A: `decile_score` (1â€“10). Model B: `two_year_recid` (binary, 1 = re-offended within two years), which is the dataset's `is_recidivist` target. `decile_score` is never used as a feature (it would leak COMPAS's own prediction).

```python
df = raw[(raw.days_b_screening_arrest.between(-30, 30)) & (raw.is_recid != -1) &
         (raw.c_charge_degree != 'O') & (raw.score_text != 'N/A')].copy().reset_index(drop=True)
print('After ProPublica filtering:', df.shape)

df['juv_total']  = df.juv_fel_count + df.juv_misd_count + df.juv_other_count
df['male']       = (df.sex == 'Male').astype(int)
df['felony']     = (df.c_charge_degree == 'F').astype(int)
df['race_grp']   = df.race.where(df.race.isin(['African-American','Caucasian','Hispanic']), 'Other')
df['race_African_American'] = (df.race_grp == 'African-American').astype(int)
df['race_Hispanic']         = (df.race_grp == 'Hispanic').astype(int)
df['race_Other']            = (df.race_grp == 'Other').astype(int)
df['y']  = df.two_year_recid.astype(int)       # Model B target
df['ds'] = df.decile_score                      # Model A target

FEATS = ['age','priors_count','juv_total','male','felony','race_African_American','race_Hispanic','race_Other']
CONT  = ['age','priors_count','juv_total']
assert df[FEATS+['y','ds']].isna().sum().sum() == 0
print('Missing values in model columns: 0')
print('Base rate of recidivism:', round(df.y.mean(),4))
print(df.race_grp.value_counts())
```

```python
# ---- Train/test split: 70/30, random_state = 210 ----
train_idx, test_idx = train_test_split(df.index, test_size=0.30, random_state=SEED)
tr, te = df.loc[train_idx].copy(), df.loc[test_idx].copy()
# outlier handling: winsorize priors_count at train 99th percentile
cap = tr.priors_count.quantile(0.99); print('priors_count 99th pct cap (train):', cap)
for part in (tr, te): part['priors_count'] = part['priors_count'].clip(upper=cap)
df['priors_count'] = df['priors_count'].clip(upper=cap)   # same cap for full-data analyses (2.1)
print('Train:', tr.shape, ' Test:', te.shape)

# scaling (fit on train only)
scaler = StandardScaler().fit(tr[CONT])
def design(d):
    X = d[FEATS].copy(); X[CONT] = scaler.transform(d[CONT]); return X
Xtr, Xte = design(tr), design(te)

# save the cleaned dataset (the 'created dataset')
df.drop(columns=['race_grp']).to_csv('compas_cleaned_S24BCAU0210.csv', index=False)
print('Saved cleaned dataset -> compas_cleaned_S24BCAU0210.csv')
```

## Part 1 â€“ Build: Model A (OLS on decile_score) vs Model B (Logistic on is_recidivist)

```python
lin = LinearRegression().fit(Xtr, tr.ds)
log = LogisticRegression(penalty=None, max_iter=5000).fit(Xtr, tr.y)

predA = lin.predict(Xte)                  # continuous decile prediction
medA  = np.median(predA)                  # threshold = median of Model A's test predictions
flagA = (predA >= medA).astype(int)       # 1 = high risk
pB    = log.predict_proba(Xte)[:,1]
flagB = (pB >= 0.5).astype(int)

print(f'Model A (OLS) test R^2 on decile_score: {lin.score(Xte, te.ds):.4f}')
print(f'Median of Model A predictions (test): {medA:.4f}   (scale 1-10)')
print(f'Model B test accuracy @0.5: {accuracy_score(te.y, flagB):.4f}, AUC: {roc_auc_score(te.y, pB):.4f}')
print(f'Share flagged high-risk -> A(median): {flagA.mean():.2%}   B(0.5): {flagB.mean():.2%}')

cm = pd.crosstab(pd.Series(flagA, name='Model A (median threshold)').map({0:'Low',1:'High'}),
                 pd.Series(flagB, name='Model B (p>=0.5)').map({0:'Low',1:'High'}))
print('\nAgreement / disagreement confusion matrix (counts):'); display(cm)
mism = int((flagA != flagB).sum())
print(f'Mismatched cases: {mism} of {len(te)} = {mism/len(te):.2%}')
print(f'  A=High & B=Low: {int(((flagA==1)&(flagB==0)).sum())}   A=Low & B=High: {int(((flagA==0)&(flagB==1)).sum())}')
print(f'Mean of Model B probabilities: {pB.mean():.4f}; median: {np.median(pB):.4f}')
print(f'Spearman rank correlation between A and B scores: {stats.spearmanr(predA, pB)[0]:.4f}')
# why they differ: where do the mismatches sit?
mm = te.assign(flagA=flagA, flagB=flagB, pB=pB, predA=predA)[flagA!=flagB]
print('\nMismatched cases - mean of key features vs test overall:')
display(pd.DataFrame({'mismatched': mm[['age','priors_count','juv_total','male','felony','pB','predA']].mean(),
                      'test_all':   te.assign(pB=pB, predA=predA)[['age','priors_count','juv_total','male','felony','pB','predA']].mean()}))
```

```python
# Extra robustness: 5-fold stratified cross-validation on the TRAIN set (the 'CV')
skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=SEED)
cvB = cross_validate(LogisticRegression(penalty=None, max_iter=5000), Xtr, tr.y, cv=skf,
                     scoring=['accuracy','roc_auc','precision','recall'])
cvA = cross_validate(LinearRegression(), Xtr, tr.ds, cv=5, scoring=['r2','neg_mean_absolute_error'])
print('Logistic 5-fold CV (train):')
for k in ['accuracy','roc_auc','precision','recall']:
    print(f'  {k}: {cvB["test_"+k].mean():.4f} +/- {cvB["test_"+k].std():.4f}')
print('OLS 5-fold CV (train):  R2 = %.4f +/- %.4f ; MAE = %.4f' % (cvA['test_r2'].mean(), cvA['test_r2'].std(), -cvA['test_neg_mean_absolute_error'].mean()))
```

## Part 2.1 â€“ Simpson's Paradox Hunt

```python
RAW = ['age','priors_count','juv_total','male','felony','race_African_American','race_Hispanic','race_Other']
def fit_logit(d, drop=()):
    cols = [c for c in RAW if c not in drop]
    return sm.Logit(d.y, sm.add_constant(d[cols])).fit(disp=0)

# Scan several candidate splits for the race_African_American coefficient (key predictor)
rows=[]
full_m = fit_logit(df)
for var, drop in [('male',('male',)), ('felony',('felony',))]:
    for v in [0,1]:
        s = df[df[var]==v]; m = fit_logit(s, drop)
        ci = m.conf_int().loc['race_African_American']
        rows.append([f'{var}={v}', len(s), m.params['race_African_American'], ci[0], ci[1], m.pvalues['race_African_American']])
scan = pd.DataFrame(rows, columns=['subgroup','n','coef_AA','CI_low','CI_high','p'])
ci = full_m.conf_int().loc['race_African_American']
scan.loc[-1] = ['FULL', len(df), full_m.params['race_African_American'], ci[0], ci[1], full_m.pvalues['race_African_American']]
print('Scan of the race_African_American coefficient across splits:'); display(scan.sort_index().round(4))
```

```python
# Chosen split: SEX. Key predictor: race_African_American (also report priors_count, age)
fem, mal = df[df.male==0], df[df.male==1]
mF, mA, mB = full_m, fit_logit(fem, ('male',)), fit_logit(mal, ('male',))
rows=[]
for name, m, n in [('Full dataset', mF, len(df)), ('Subgroup A: Female', mA, len(fem)), ('Subgroup B: Male', mB, len(mal))]:
    for k in ['race_African_American','priors_count','age']:
        c = m.conf_int().loc[k]
        rows.append([name, n, k, m.params[k], c[0], c[1], m.pvalues[k], np.exp(m.params[k])])
t21 = pd.DataFrame(rows, columns=['model','n','predictor','coef','CI_low','CI_high','p_value','odds_ratio']).round(4)
display(t21)

# test whether Female vs Male coefficients differ (z-test)
for k in ['race_African_American','priors_count']:
    z = (mA.params[k]-mB.params[k]) / np.sqrt(mA.bse[k]**2 + mB.bse[k]**2)
    print(f'{k}: Female {mA.params[k]:.4f} vs Male {mB.params[k]:.4f} -> z = {z:.3f}, p = {2*(1-stats.norm.cdf(abs(z))):.4f}')
print('Sign reversal for race_African_American (female vs full)?', np.sign(mA.params['race_African_American']) != np.sign(mF.params['race_African_American']))
print('\nRecidivism rate by sex x race (supports the explanation):')
display(df.groupby(['male','race_African_American']).agg(n=('y','size'), recid_rate=('y','mean'), mean_priors=('priors_count','mean'), mean_age=('age','mean')).round(3))
```

## Part 2.2 â€“ Omitted-Variable Bias (drop `priors_count`)

```python
DROPPED = 'priors_count'
fm = fit_logit(tr); rm = fit_logit(tr, (DROPPED,))
common = [c for c in RAW if c != DROPPED]
t22 = pd.DataFrame({'full_coef': fm.params[common], 'reduced_coef': rm.params[common]})
t22['abs_change'] = t22.reduced_coef - t22.full_coef
t22['pct_change'] = 100*(t22.reduced_coef - t22.full_coef)/t22.full_coef.abs()
t22['full_p'] = fm.pvalues[common]; t22['reduced_p'] = rm.pvalues[common]
display(t22.round(4))
top = t22['pct_change'].abs().idxmax()
print(f'Dropped variable: {DROPPED}.  Largest shift: {top} ({t22.loc[top,"pct_change"]:.1f}%)')
print('Correlation of priors_count with the other predictors (why it confounds):')
display(tr[RAW].corr()[DROPPED].drop(DROPPED).round(3))
print('Test AUC full vs reduced:',
      round(roc_auc_score(te.y, fm.predict(sm.add_constant(te[RAW]))),4), round(roc_auc_score(te.y, rm.predict(sm.add_constant(te[common]))),4))
```

## Part 2.3 â€“ Adversarial Subset Construction

```python
# Logistic model from Part 1 (scaled features); test-set probabilities pB
te['pB'] = pB
fp = te[(te.pB >= 0.75) & (te.y == 0)]
fn = te[(te.pB <= 0.25) & (te.y == 1)]
bad = pd.concat([fp.assign(kind='confident FP'), fn.assign(kind='confident FN')])
print(f'Confident wrong predictions: {len(bad)} of {len(te)} = {len(bad)/len(te):.2%}  (FP: {len(fp)}, FN: {len(fn)})')
adv = bad if len(bad) <= 50 else bad.sample(50, random_state=SEED)
print('Adversarial subset size:', len(adv), '|', adv.kind.value_counts().to_dict())

def profile(d):
    return pd.Series({'mean_age': d.age.mean(), 'median_age': d.age.median(), 'pct_age<25': (d.age<25).mean()*100,
        'pct_age>45': (d.age>45).mean()*100, 'pct_male': d.male.mean()*100, 'pct_African_American': d.race_African_American.mean()*100,
        'pct_felony': d.felony.mean()*100, 'mean_priors': d.priors_count.mean(), 'pct_priors==0': (d.priors_count==0).mean()*100,
        'pct_priors>=5': (d.priors_count>=5).mean()*100, 'pct_juv>0': (d.juv_total>0).mean()*100})
prof = pd.DataFrame({'adversarial_50': profile(adv), 'full_test': profile(te), 'all_confident_wrong': profile(bad)})
prof['diff (adv - test)'] = prof.adversarial_50 - prof.full_test
display(prof.round(2))
print('Model coefficients (scaled units):'); display(pd.Series(log.coef_[0], index=FEATS).round(4))
print('Mean predicted prob in subset:', adv.pB.mean().round(3), '| actual recid rate in subset:', adv.y.mean().round(3))
print('Within the subset, the FP rows vs FN rows:'); display(adv.groupby('kind')[['age','priors_count','juv_total','male','felony','race_African_American','pB']].mean().round(3))
# Check an interaction as candidate explanation: priors x age
inter = sm.Logit(tr.y, sm.add_constant(tr[RAW].assign(priors_x_age=tr.priors_count*tr.age))).fit(disp=0)
print('Interaction priors_count x age: coef = %.5f, p = %.4g' % (inter.params['priors_x_age'], inter.pvalues['priors_x_age']))
```

## Part 2.4 â€“ Fairness Trade-Off Analysis (Group A = African-American, Group B = Caucasian)

```python
A = te[te.race_African_American==1]; B = te[(te.race_African_American==0) & (te.race_Hispanic==0) & (te.race_Other==0)]
print('Group A (African-American) n =', len(A), '| Group B (Caucasian) n =', len(B))
print('Base recid rate A: %.3f  B: %.3f' % (A.y.mean(), B.y.mean()))

def rates(g, t):
    yhat = (g.pB >= t).astype(int); y = g.y
    tp=((yhat==1)&(y==1)).sum(); fp_=((yhat==1)&(y==0)).sum(); fn_=((yhat==0)&(y==1)).sum(); tn=((yhat==0)&(y==0)).sum()
    return dict(sel=yhat.mean(), tpr=tp/(tp+fn_) if tp+fn_ else np.nan, fpr=fp_/(fp_+tn) if fp_+tn else np.nan,
                ppv=tp/(tp+fp_) if tp+fp_ else np.nan)
def ratio(a,b): return a/b if (b not in (0,np.nan) and b==b and b!=0) else np.nan

thr = np.round(np.arange(0.05, 0.96, 0.10), 2); rows=[]
for t in thr:
    ra, rb = rates(A,t), rates(B,t)
    dp = ratio(ra['sel'], rb['sel']); eo = np.nanmin([ratio(ra['tpr'],rb['tpr']), ratio(ra['fpr'],rb['fpr'])]); pp = ratio(ra['ppv'], rb['ppv'])
    allp = pd.concat([A,B]); acc = accuracy_score(allp.y, (allp.pB>=t).astype(int))
    rows.append(dict(threshold=t, DP=dp, EO=eo, PP=pp, TPR_A=ra['tpr'], TPR_B=rb['tpr'], FPR_A=ra['fpr'], FPR_B=rb['fpr'],
                     sel_A=ra['sel'], sel_B=rb['sel'], PPV_A=ra['ppv'], PPV_B=rb['ppv'], accuracy=acc))
fair = pd.DataFrame(rows); display(fair.round(4))
```

```python
fig, ax = plt.subplots(figsize=(7.5,6.5))
ax.plot(fair.DP, fair.EO, '-o', color='tab:blue', alpha=.6)
for _, r in fair.iterrows(): ax.annotate(f't={r.threshold:.2f}', (r.DP, r.EO), textcoords='offset points', xytext=(6,6), fontsize=9)
ax.axvline(1, ls='--', c='grey'); ax.axhline(1, ls='--', c='grey'); ax.scatter([1],[1], marker='*', s=200, c='gold', edgecolor='k', zorder=5, label='Ideal (1,1)')
ax.set_xlim(0,1.5); ax.set_ylim(0,1.5); ax.set_xlabel('Demographic Parity ratio (A/B)'); ax.set_ylabel('Equalized Odds  min(TPR ratio, FPR ratio)')
ax.set_title('Fairness trade-off across thresholds (COMPAS, AA vs Caucasian)'); ax.legend(); ax.grid(alpha=.3)
plt.tight_layout(); plt.savefig('fairness_tradeoff.png', dpi=200); plt.show()

# also: all three metrics vs threshold
fig, ax = plt.subplots(figsize=(8,4.5))
for m in ['DP','EO','PP']: ax.plot(fair.threshold, fair[m], '-o', label=m)
ax.axhline(1, ls='--', c='grey'); ax.set_xlabel('threshold'); ax.set_ylabel('ratio (ideal = 1)'); ax.legend(); ax.grid(alpha=.3)
plt.tight_layout(); plt.savefig('fairness_vs_threshold.png', dpi=200); plt.show()
```

```python
# Quantify the trade-off. t=0.05 and t=0.95 are degenerate (everyone / almost no one flagged), so 'best' picks use 0.15-0.85.
f = fair[(fair.threshold>=0.15)&(fair.threshold<=0.85)].copy()
for m in ['DP','EO','PP']: f[m+'_gap'] = (f[m]-1).abs()
tDP, tEO, tPP = f.loc[f.DP_gap.idxmin(),'threshold'], f.loc[f.EO_gap.idxmin(),'threshold'], f.loc[f.PP_gap.idxmin(),'threshold']
print('Threshold closest to ideal (0.15-0.85) -> DP:', tDP, '| EO:', tEO, '| PP:', tPP)
fi = fair.set_index('threshold')
def tradeoff(t_from, t_to, gain, loss):
    a, b = fi.loc[t_from], fi.loc[t_to]
    g = (abs(b[gain]-1)-abs(a[gain]-1))/abs(a[gain]-1)*100      # % change in distance-from-ideal (negative = improvement)
    l = (abs(b[loss]-1)-abs(a[loss]-1))/abs(a[loss]-1)*100
    print(f't={t_from} -> t={t_to}: {gain} {a[gain]:.3f}->{b[gain]:.3f} (gap {g:+.1f}%) | {loss} {a[loss]:.3f}->{b[loss]:.3f} (gap {l:+.1f}%) | '
          f'accuracy {a.accuracy:.3f}->{b.accuracy:.3f}')
    if g != 0: print(f'    trade-off cost = {abs(l/g):.2f}  (change in {loss} gap per unit change in {gain} gap)')
print('NOTE: here DP and EO ratios are both >1 and move together, so DP vs EO is NOT the real conflict; DP/EO vs PP and accuracy is.')
tradeoff(0.65, 0.25, 'DP', 'PP')   # lowering threshold: DP improves, PP gets worse
tradeoff(0.55, 0.25, 'DP', 'PP')
tradeoff(0.45, 0.25, 'DP', 'EO')
tradeoff(0.55, 0.35, 'EO', 'PP')
gaps = f[['DP_gap','EO_gap','PP_gap']].max(axis=1)
print('\nBest worst-case threshold: min over t of max(|DP-1|,|EO-1|,|PP-1|) =', gaps.min().round(4), 'at t =', f.loc[gaps.idxmin(),'threshold'])
```

# Part 3 â€” Descriptive Analysis Questions

Each question below directly quotes and explains the empirical results computed in Parts 1 and 2 of this notebook.

### Question 1: Linear Regression Threshold vs Direct Logistic Regression Classification

The threshold-based linear regression approach (Model A) and direct logistic regression approach (Model B) exhibited substantial classification discordance, producing **237 mismatched cases out of 1,852 test instances**â€”an overall mismatch rate of exactly **12.80%**. Crucially, this disagreement is heavily one-directional: **222 individuals** were flagged as high risk by Model A while being designated low risk by Model B, whereas only **15 individuals** were flagged as low risk by Model A and high risk by Model B. This structural divergence occurs because the two methodologies operationalize classification through fundamentally incompatible mathematical mechanisms.

Model A applies a relative, rank-ordering threshold fixed at the test median of its continuous predictions (**4.4972** on the 1-to-10 decile scale). By definition, thresholding at the sample median mechanically mandates that exactly **50.05%** of the test cohort is labeled high risk, irrespective of whether the underlying base rate warrants that volume of positive flags. In contrast, Model B imposes an absolute probabilistic cutoff at (\text{recidivism}) \ge 0.5$. Within our test cohort, the mean predicted probability of recidivism is **0.4568** and the median is **0.4473**, reflecting an empirical base recidivism rate of **45.51%**. Consequently, Model B's 0.5 threshold requires an individual to sit above the **61.12th percentile** of the risk distribution, flagging only **38.88%** of defendants.

The 222 individuals flagged high-risk exclusively by Model A occupy the probability band between approximately 0.42 and 0.49. Examining their demographic profile reveals they are significantly younger than the test average (mean age **26.73** versus **34.84**) and possess low prior convictions (mean priors **1.52** versus **3.25**). Model A's linear formulation heavily penalizes youth, pushing predicted decile scores just above 4.4972, whereas Model B's sigmoid function dampens these intermediate scores, correctly retaining them below the 0.5 certainty threshold.

### Question 2: Simpson's Paradox Subgroup Discovery and Policy Implications

Auditing across demographic subgroups uncovered a textbook manifestation of Simpson's Paradox when partitioning by biological sex. In the pooled logistic model fitted on the entire dataset of 6,172 individuals, the coefficient for ace_African_American is $\beta = +0.0539$ with a 95% confidence interval of $[-0.0700, +0.1778]$ and a p-value of **0.3942**. In this aggregate specification, race appears statistically indistinguishable from zero, suggesting to an uncritical analyst that racial identity exerts no independent influence on recidivism once controlling for age, prior convictions, and felony status.

However, disaggregating the data by sex completely inverts this relationship. For the female subgroup (Subgroup A,  = 1,175$), the coefficient on ace_African_American becomes negative and statistically significant: **$\beta = -0.3197$**, 95% CI $[-0.6047, -0.0348]$,  = 0.0279$, corresponding to an odds ratio of **0.7264**. For women, identifying as African-American is associated with an estimated **27.36% reduction in the odds of recidivism** relative to Caucasian women with identical criminal histories. Conversely, for the male subgroup (Subgroup B,  = 4,997$), the racial coefficient is positive and statistically significant: **$\beta = +0.1466$**, 95% CI $[+0.0085, +0.2848]$,  = 0.0375$, corresponding to an odds ratio of **1.1579** (a **15.79% increase** in recidivism odds). ### Question 4: Practitioner Methodology for Detecting Hidden Confounds in the Real World

To detect hidden, unobserved confounds in real-world algorithmic systems where omitted variables cannot be directly inspected, practitioners cannot rely on standard goodness-of-fit metrics. I propose a rigorous, three-stage auditing framework combining negative control outcomes, sensitivity bounds, and external temporal validation.

The primary technical component is the implementation of **Negative Control Outcomes (NCOs)**. In epidemiology and econometrics, an NCO is a secondary outcome variable that is causally known to be unaffected by the true underlying risk mechanism but is exposed to the identical suspected confounding environment (e.g., systemic policing patterns or socioeconomic deprivation). In criminal risk assessment, an NCO could be non-criminal civil summonses, procedural court appearance compliance, or traffic infractions occurring prior to adjudication. If our predictive featuresâ€”specifically race or neighborhood deprivationâ€”predict the negative control outcome with statistical significance despite having no causal connection to it, the model is definitively contaminated by unobserved confounding.

Second, practitioners must compute **Oster's delta** (Oster, 2019) and Rosenbaum sensitivity bounds. This assesses how strongly an unobserved confounder would need to correlate with both the treatment/demographic feature and the recidivism outcome relative to observed covariates to overturn the estimated model coefficients. If an unobserved factor with only 20% of the explanatory power of prior criminal history could flip the sign or significance of a sensitive coefficient, the model must be deemed fragile.

Third, models must undergo external spatial-temporal validation across jurisdictions possessing divergent policing regimes (e.g., contrasting Broward County with a county utilizing non-custodial citation models). Divergent coefficient behavior across jurisdictions immediately diagnoses the presence of local unobserved confounds embedded in administrative arrest data.
